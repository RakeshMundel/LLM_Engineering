```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
import gradio as gr
```

```python
load_dotenv(override=True)
gem_api_key=os.getenv("GEMINI_API_KEY")

if gem_api_key:
    print(f"gemini api key exists and begin with {gem_api_key[:3]}...")
else:
    print("gemini api key not found")
```

```python
gemini=OpenAI(api_key=gem_api_key,base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
system_message="You are a helpful assistant"

def message_gem(prompt):
    messages=[{"role":"system","content":system_message},{"role":"user","content":prompt}]
    response=gemini.chat.completions.create(model="model="gemini-3.1-flash-lite"",messages=messages)
    return response.choices[0].message.content
```

```python
print(message_gem("What is artificial intelligence?"))
```

```python
message_gem("what is today's date?")
```

```python
system_message = "You are a helpful assistant that responds in markdown without code blocks"

message_input = gr.Textbox(label="Your message:", info="Enter a message for GPT-4.1-mini", lines=7)
message_output = gr.Textbox(label="Response:")

view = gr.Interface(
    fn=message_gem,
    title="GPT", 
    inputs=[message_input], 
    outputs=[message_output], 
    examples=[
        "Explain the Transformer architecture to a layperson",
        "Explain the Transformer architecture to an aspiring AI engineer",
        ], 
    flagging_mode="never"
    )
view.launch()
```

```python
def stream_gpt(prompt):
    messages=[
        {"role":"system","content":system_message},
        {"role":"user","content":prompt}
    ]

    stream=gemini.chat.completions.create(
        model='gemini-3.7-flash',
        messages=messages,
        stream=True
    )
    result=""
    for chunk in stream:
        result+=chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input=gr.Textbox(label="Your message:",info="Enter a message for gemini",lines=7)
message_output=gr.Markdown(label="Response:")

view=gr.Interface(
    fn=stream_gpt,
    inputs=[message_input],
    outputs=[message_output],
    examples=[
        "Explain the Transformer architecture",
        "Expain the panda library"
    ],
    flagging_mode="never"
)
view.launch()
```
